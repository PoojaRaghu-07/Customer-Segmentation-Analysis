```python
import matplotlib.pyplot as plt
import matplotlib.image as mpimg
from matplotlib.gridspec import GridSpec
from pathlib import Path

# =====================================================
# CUSTOMER VALUE SEGMENTATION DASHBOARD
# =====================================================

folder = Path("data visualization")
charts = sorted(folder.glob("*.png"))

if len(charts) == 0:
    raise FileNotFoundError(
        "No PNG files found inside 'data visualization' folder."
    )

# =====================================================
# CREATE FIGURE
# =====================================================

fig = plt.figure(figsize=(24, 34))
fig.patch.set_facecolor("#F8F9FA")

# 5 rows:
# Row 0-3 : Charts
# Row 4   : Insights

gs = GridSpec(
    5,
    3,
    figure=fig,
    height_ratios=[1, 1, 1, 1, 0.9],
    hspace=0.28,
    wspace=0.08
)

# =====================================================
# DASHBOARD TITLE
# =====================================================

fig.suptitle(
    "📊 Customer Value Segmentation Dashboard\n"
    "Customer Analytics using RFM Analysis",
    fontsize=30,
    fontweight="bold",
    y=0.985
)

# =====================================================
# PROJECT WORKFLOW CARDS
# =====================================================

card_style = dict(
    boxstyle="round,pad=0.6",
    edgecolor="black",
    linewidth=1.5
)

plt.figtext(
    0.13,
    0.935,
    "🧹\nData\nCleaning",
    ha="center",
    fontsize=14,
    fontweight="bold",
    bbox=dict(facecolor="#D6EAF8", **card_style)
)

plt.figtext(
    0.37,
    0.935,
    "📈\nExploratory\nData Analysis",
    ha="center",
    fontsize=14,
    fontweight="bold",
    bbox=dict(facecolor="#D5F5E3", **card_style)
)

plt.figtext(
    0.63,
    0.935,
    "🎯\nRFM\nSegmentation",
    ha="center",
    fontsize=14,
    fontweight="bold",
    bbox=dict(facecolor="#FCF3CF", **card_style)
)

plt.figtext(
    0.87,
    0.935,
    "📊\nExecutive\nDashboard",
    ha="center",
    fontsize=14,
    fontweight="bold",
    bbox=dict(facecolor="#FADBD8", **card_style)
)

# =====================================================
# CHARTS
# =====================================================

axes = []

for i in range(4):
    for j in range(3):
        ax = fig.add_subplot(gs[i, j])
        axes.append(ax)

for ax, chart in zip(axes, charts):

    img = mpimg.imread(chart)

    ax.imshow(img)

    ax.axis("off")

for ax in axes[len(charts):]:
    ax.axis("off")

# =====================================================
# INSIGHTS
# =====================================================

insights = """
📌 EXECUTIVE BUSINESS INSIGHTS

✔ Successfully completed an end-to-end Customer Value Segmentation project
covering Data Cleaning, Data Quality Assessment, Feature Engineering,
Exploratory Data Analysis, RFM Segmentation and Dashboard Visualization.

✔ RFM analysis successfully classified customers into meaningful business
segments such as Champions, Loyal, Potential, At-Risk and Lost Customers.

✔ Champions and Loyal Customers represent the highest-value customer groups
and should be prioritized through loyalty programs and personalized offers.

✔ Frequency and Monetary Value exhibit a strong positive relationship,
indicating repeat customers contribute significantly to overall revenue.

✔ Customer demographic analysis including Age, Gender, Wealth Segment,
Job Industry and Car Ownership provides deeper understanding of customer behaviour.

✔ The dashboard integrates demographic and transactional insights,
supporting business intelligence and data-driven decision making.

✔ This project demonstrates a complete analytics workflow from raw data
to actionable business insights.
"""

insight_ax = fig.add_subplot(gs[4, :])

insight_ax.axis("off")

insight_ax.text(
    0.5,
    0.5,
    insights,
    ha="center",
    va="center",
    fontsize=15,
    fontweight="bold",
    wrap=True,
    linespacing=1.8,
    bbox=dict(
        facecolor="#F8F9FA",
        edgecolor="#444444",
        linewidth=2,
        boxstyle="round,pad=1"
    )
)

# =====================================================
# LAYOUT
# =====================================================

plt.subplots_adjust(
    top=0.90,
    bottom=0.03,
    left=0.03,
    right=0.97
)

# =====================================================
# SAVE
# =====================================================

plt.savefig(
    "Customer_Segmentation_Dashboard.png",
    dpi=300,
    bbox_inches="tight",
    facecolor=fig.get_facecolor()
)

plt.show()
```
